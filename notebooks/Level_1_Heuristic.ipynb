# 🎯 Level 1: Heuristic Prediction (Hand-Written Rule)

Welcome to the **Seoul Apartment Price Prediction** workshop! 
In this first level, we will create a simple prediction model using **intuition (heuristics)**, not complex AI.

**Goal**: Predict apartment prices based on **District** (Gu) and **Area** (Space).
**Our Rule**: "The price is the **Average (Median) Price** of that district."

### 1. Load Data
We will load the dataset. This works in Google Colab by downloading the file from GitHub.

```python
import pandas as pd
import io
import requests

# Try to load local file, otherwise download from GitHub
try:
    df = pd.read_csv('../data/sample_apt_price_data.csv')
    print("Loaded from local file!")
except FileNotFoundError:
    print("Local file not found. Downloading from GitHub...")
    url = "https://raw.githubusercontent.com/bookseal/seoul-apt-price-prediction/main/data/sample_apt_price_data.csv"
    content = requests.get(url).content
    df = pd.read_csv(io.StringIO(content.decode('utf-8')))
    print("Loaded from GitHub!")

# Show first 5 rows
df.head()
```

### 2. Explore Data (EDA)
Let's see what columns we have.
*   `district`: The district name (e.g., Gangnam-gu)
*   `area_m2`: The size of the apartment
*   `price_10k_krw`: The price in 10,000 KRW (e.g., 10000 = 100 Million KRW)

```python
df.info()
```

### 3. Create our "Model" (The Logic)
We want to calculate the **Median Price** for each district.
Wait, why Median and not Mean (Average)?
*   **Mean**: Can be distorted by super expensive luxury apartments (Outliers).
*   **Median**: The "middle" price, which is more representative for typical apartments.

```python
# Group by District and calculate Median Price
district_median = df.groupby('district')['price_10k_krw'].median()

# Sort from expensive to cheap
district_median = district_median.sort_values(ascending=False)

print("Most Expensive Districts (Average Price):")
print(district_median.head())
```

### 4. Making Predictions
Now we can create a `predict_price` function.
If you ask for "Gangnam-gu", it returns the Gangnam median.
If you ask for "Nowon-gu", it returns the Nowon median.
**Note**: We ignore `area` for now (that's the limitation of this level!).

```python
def predict_price(district_name, area_m2):
    # Get the median price for the district
    if district_name in district_median:
        prediction = district_median[district_name]
        return prediction
    else:
        return "District not found!"

# Test it!
gangnam_price = predict_price("강남구", 84)
nowon_price = predict_price("노원구", 84)

print(f"Predicted price for Gangnam 84m2: {gangnam_price} (10k KRW)")
print(f"Predicted price for Nowon 84m2:   {nowon_price} (10k KRW)")
```
